Вам надо смоделировать шахматную доску, которая представляет собой numpy-матрицу. Чёрная 
клетка представляется нулём, а белая — единицей. Если смотреть на доску сверху, то левая верхняя

Вы уже практически реализовали задачу, ведь вы знаете, что у шахматной доски 8 столбцов и 8 
строк, но заказчик вдруг изменил требования и заявил, что в его игре доска может быть 
произвольного размера, но всегда квадратной, например, 50х50.  
Напишите функцию make_field(size), в которую поступает размер доски, и которая возвращает 
требуемый массив numpy. Тип элементов массива должен быть np.int8.

```python
def make_field(size):
    return ((np.indices((size, size)).sum(axis=0) % 2).astype(np.int8))

print("Поле 5x5:\n", make_field(5))
print("Поле 10x10:\n", make_field(10))
```

№3 (1 балл)  
Напишите функцию super_sort(rows, cols), в которую поступают размеры матрицы. Вам 
необходимо:  
  
- заполнить матрицу А случайными целыми числами из диапазона [1..100] 
- сделать копию этой матрицы в матрицу B 
- упорядочить чётные столбцы матрицы B по возрастанию, а нечётные — по убыванию  
  
Верните кортеж, состоящий из матриц А и B.  
  
Для решения задачи нельзя использовать циклы. Для сортировки используйте функцию np.sort.  

```python
def super_sort(rows, cols):
    A = np.random.randint(1, 101, (rows, cols))
    B = A.copy()

    even_cols = np.arange(0, cols, 2)
    B[:, even_cols] = np.sort(B[:, even_cols], axis=0)

    odd_cols = np.arange(1, cols, 2)
    B[:, odd_cols] = np.sort(B[:, odd_cols], axis=0)[::-1]
    return A, B

A, B = super_sort(4, 5)
print("A:\n", A)
print("B:\n", B)
```

№5 (2 балла)  
Создайте две numpy матрицы размера nхn и заполните их случайными значениями. Посчитайте их 
поэлементное и матричное произведения. Для этого реализуйте функции:  
1. mult – матричное перемножение без использования numpy 
2. el_mult – поэлементное перемножение без использования numpy 
3. np_mult – матричное перемножение c использованием numpy 
4. np_el_mult – поэлементное перемножение с использованием numpy  
  
При помощи timeit засеките время работы функций для n = 100000 (для чистоты эксперимента 
засекайте время на одинаковых матрицах). Во сколько раз numpy версия функций работает быстрее 
реализаций без numpy?

```python
import timeit

def mult(A, B):
    n = len(A)
    result = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            s = 0
            for k in range(n):
                s += A[i][k] * B[k][j]
            result[i][j] = s
    return result

def el_mult(A, B):
    n = len(A)
    result = [[A[i][j] * B[i][j] for j in range(n)] for i in range(n)]
    return result

def np_mult(A, B):
    return np.dot(A, B)

def np_el_mult(A, B):
    return A * B

n = 400

A_np = np.random.randint(0, 10, (n, n))
B_np = np.random.randint(0, 10, (n, n))
A_list = A_np.tolist()
B_list = B_np.tolist()

t1 = timeit.timeit(lambda: mult(A_list, B_list), number=1)
t2 = timeit.timeit(lambda: el_mult(A_list, B_list), number=1)
t3 = timeit.timeit(lambda: np_mult(A_np, B_np), number=1)
t4 = timeit.timeit(lambda: np_el_mult(A_np, B_np), number=1)

print(f"mult (без numpy): {t1:.4f} сек")
print(f"el_mult (без numpy): {t2:.4f} сек")
print(f"np_mult (numpy): {t3:.4f} сек")
print(f"np_el_mult (numpy): {t4:.4f} сек")

print("Матричное умножение numpy быстрее в", round(t1/t3, 2), "раз(а)")
print("Поэлементное умножение numpy быстрее в", round(t2/t4, 2), "раз(а)")
```

Напишите программу, которая будет решать систему линейных уравнений вида:  
Коэффициенты и правую часть вводит пользователь. Программа выводит решение системы. Для 
нахождения решения используйте numpy.linalg.solve.   
Система не имеет решения (или имеет их бесконечно много), если определитель матрицы 
коэффициентов равен 0. В этом случае необходимо выводить соответствующее сообщение. Для 
нахождения определителя используйте функцию numpy.linalg.det.

```python
a11 = float(input("Введите коэффициент a11: "))
a12 = float(input("Введите коэффициент a12: "))
a21 = float(input("Введите коэффициент a21: "))
a22 = float(input("Введите коэффициент a22: "))
b1 = float(input("Введите правую часть b1: "))
b2 = float(input("Введите правую часть b2: "))

A = np.array([[a11, a12],
              [a21, a22]])

b = np.array([b1, b2])

det = np.linalg.det(A)

if np.isclose(det, 0):
    print("Система не имеет единственного решения (определитель равен 0).")
else:
    solution = np.linalg.solve(A, b)
    print("Решение системы:", solution)
```

Используя numpy, без питоновских циклов сконвертируйте цветное изображение в чёрно-белое по 
формуле:  
  
C = 0.2989 * R + 0.5870 * G + 0.1140 * B  
  
Считается, что вычисленный таким образом цвет лучше всего подходит для человеческого глаза, 
поскольку яркость каждой компоненты глаз воспринимает по-разному.  
  
Напишите функцию bw_convert(), в которой проведите все необходимые операции.  
Не забудьте, что значение каждой составляющей цвета — целое число, поэтому результат 
необходимо округлить. Для этого воспользуйтесь методом round() из библиотеки numpy.   
Используйте формат изображения RGB и библиотеку pillow.

```python
from PIL import Image
import numpy as np

def bw_convert(image_path, save_path=None):
    img = Image.open(image_path).convert('RGB')
    arr = np.array(img)
    bw = np.round(0.2989 * arr[..., 0] + 0.5870 * arr[..., 1] + 0.1140 * arr[..., 2]).astype(np.uint8)
    bw_img = Image.fromarray(bw, mode='L')
    if save_path:
        bw_img.save(save_path)
    return bw_img

bw_img = bw_convert('input.jpg', 'output_bw.jpg')
bw_img.show()
```
