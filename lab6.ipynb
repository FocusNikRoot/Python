Напишите класс Balance для описания весов с двумя чашами. На левую и правую чашу объекта будут 
добавляться грузы с различным весом, ваша задача определить положение чаш.  
Метод add_right принимает целое число — вес, положенный на правую чашу весов, add_left — на 
левую чашу. Метод result должен возвращать символ =, если вес на чашах одинаковый, R — если 
перевесила правая, L — если перевесила левая.

```python
class Balance:
    def __init__(self):
        self.left = 0
        self.right = 0

    def add_left(self, weight):
        self.left += weight

    def add_right(self, weight):
        self.right += weight

    def result(self):
        if self.left == self.right:
            return '='
        elif self.left > self.right:
            return 'L'
        else:
            return 'R'
        
balance = Balance()
balance.add_right(10)
balance.add_left(5)
balance.add_left(5)
print(balance.result())
balance.add_left(1)
print(balance.result())
```

№3 (1 балл)  
Напишите класс Selector. Экземпляр этого класса при инициализации получает список чисел. Вызов 
метода get_odds возвращает нечётные числа из первоначального списка, вызов get_evens — 
чётные.  
Числа должны идти в том же порядке, в котором они были в изначальном списке. 


```python
class Selector:
    def __init__(self, numbers):
        self.numbers = numbers

    def get_odds(self):
        return [x for x in self.numbers if x % 2 != 0]

    def get_evens(self):
        return [x for x in self.numbers if x % 2 == 0]

# Пример использования:
selector = Selector([1, 2, 3, 4, 5, 6])
print(selector.get_odds())
print(selector.get_evens())
```

Определите класс Point. При инициализации экземпляру передаются координаты x и y.  
При сравнении двух экземпляров оператор == должен возвращать True, если координаты точек 
равны, и False — если нет.  
При сравнении оператором != должно возвращаться True, если координаты точек не равны, и False 
— если равны.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __ne__(self, other):
        return not self.__eq__(other)

p1 = Point(1, 2)
p2 = Point(5, 6)

print("Equal True" if p1 == p2 else "Equal False")
print("Not Equal True" if p1 != p2 else "Not Equal False")
```

Периодически в некоторых задачах возникает надобность в разреженных массивах. Разреженный 
массив характеризуется тем, что подавляющее большинство значений в нём равны нулю, поэтому 
можно хранить только ненулевые значения. Это позволяет создавать разреженные массивы очень 
большого размера без лишних затрат памяти.  
Вам необходимо реализовать класс разреженного массива SparseArray.  
arr = SparseArray() — создание пустого разреженного массива.  
arr[i] = value — записать значение в массив.  
arr[i] — прочитать значение из массива. 

```python
class SparseArray:
    def __init__(self):
        self.data = {}

    def __setitem__(self, index, value):
        if value != 0:
            self.data[index] = value

    def __getitem__(self, index):
        return self.data.get(index, 0)

# Пример использования:
arr = SparseArray()
arr[1] = 10
arr[8] = 20
for i in range(10):
    print('arr[{}] = {}'.format(i, arr[i]))
```

```python
class Polynomial:
    def __init__(self, coefficients):
        self.coefficients = list(coefficients)

    def __call__(self, x):
        result = 0
        for power, coef in enumerate(self.coefficients):
            result += coef * (x ** power)
        return result

poly = Polynomial([0, 0, 1])
print(poly(-2))
print(poly(-1))
print(poly(0))
print(poly(1))
print(poly(2))
```

Реализуйте класс Queue, моделирующий работу очереди (принцип FIFO (First In First Out)). Очередь 
может быть инициализирована различным числом элементов.

```python
class Queue:
    def __init__(self, *values):
        self._data = list(values)

    def append(self, *values):
        self._data.extend(values)

    def copy(self):
        return Queue(*self._data)

    def pop(self):
        if not self._data:
            return None
        return self._data.pop(0)

    def extend(self, queue):
        self._data.extend(queue._data)

    def next(self):
        return Queue(*self._data[1:]) if len(self._data) > 1 else Queue()

    def __add__(self, other):
        return Queue(*(self._data + other._data))

    def __iadd__(self, other):
        self._data.extend(other._data)
        return self

    def __eq__(self, other):
        return self._data == other._data

    def __str__(self):
        return '[' + ' -> '.join(map(str, self._data)) + ']'

    def __rshift__(self, n):
        if n >= len(self._data):
            return Queue()
        return Queue(*self._data[n:])

q1 = Queue(1, 2, 3)
print(q1)
q1.append(4, 5)
print(q1)
qx = q1.copy()
print(qx.pop())
print(qx)
q2 = q1.copy()
print(q2)
print(q1 == q2, id(q1) == id(q2))
q3 = q2.next()
print(q1, q2, q3, sep='\n')
print(q1 + q3)
q3.extend(Queue(1, 2))
print(q3)
q4 = Queue(1, 2)
q4 += q3 >> 4
print(q4)
q5 = q4.next()
print(q4)
print(q5)
```

Конструктор этого класса должен принимать на вход 3 числа (длины сторон треугольника).  
Также у этого класса должен быть метод perimeter(), возвращающий периметр.  
Отнаследуйте от класса Triangle класс EquilateralTriangle, представляющий равносторонний 
треугольник.  
Переопределите конструктор этого класса с использованием конструктора базового класса.  
Конструктор класса EquilateralTriangle должен принимать на вход 1 число (длину стороны).  

```python
class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def perimeter(self):
        return self.a + self.b + self.c

class EquilateralTriangle(Triangle):
    def __init__(self, side):
        super().__init__(side, side, side)

# Пример использования:
t = Triangle(3, 4, 5)
print(t.perimeter())

et = EquilateralTriangle(6)
print(et.perimeter())
```

```python
class Weapon:
    def __init__(self, name, damage, max_dist):
        self.name = name
        self.damage = damage
        self.max_dist = max_dist

    def hit(self, target):
        dist = ((target.x - 0) ** 2 + (target.y - 0) ** 2) ** 0.5 if hasattr(self, 'owner') else 0
        
        if hasattr(self, 'owner'):
            dist = ((target.x - self.owner.x) ** 2 + (target.y - self.owner.y) ** 2) ** 0.5
        if dist > self.max_dist:
            print("Враг слишком далеко для оружия", self.name)
            return
        print(f"Врагу нанесен урон оружием {self.name} {self.damage}")
        target.get_damage(self.damage)

class BaseCharacter:
    def __init__(self, x, y, hp):
        self.x = x
        self.y = y
        self.hp = hp

    def move(self, delta_x, delta_y):
        self.x += delta_x
        self.y += delta_y

    def is_alive(self):
        return self.hp > 0

    def get_damage(self, amount):
        self.hp -= amount
        if self.hp < 0:
            self.hp = 0

    def get_coords(self):
        return (self.x, self.y)

class BaseEnemy(BaseCharacter):
    def __init__(self, x, y, weapon=None, hp=100):
        super().__init__(x, y, hp)
        self.weapon = weapon

    def hit(self, target):
        if not isinstance(target, MainHero):
            print("Могу ударить только Главного героя")
            return
        if not self.weapon:
            print("Я безоружен")
            return
        self.weapon.owner = self
        self.weapon.hit(target)

class MainHero(BaseCharacter):
    def __init__(self, x, y, name, hp):
        super().__init__(x, y, hp)
        self.name = name
        self.weapons = []
        self.current_weapon = None
        self.weapon_index = -1

    def add_weapon(self, weapon):
        self.weapons.append(weapon)
        if not self.current_weapon:
            self.current_weapon = weapon
            self.weapon_index = 0
            print(f"Подобрал {weapon.name}")
        else:
            print(f"Подобрал {weapon.name}")

    def next_weapon(self):
        if not self.weapons:
            print("Я безоружен")
            return
        self.weapon_index = (self.weapon_index + 1) % len(self.weapons)
        self.current_weapon = self.weapons[self.weapon_index]
        print(f"Сменил оружие на {self.current_weapon.name}")

    def hit(self, target):
        if not isinstance(target, BaseEnemy):
            print("Могу ударить только врага")
            return
        if not self.current_weapon:
            print("Я безоружен")
            return
        if not target.is_alive():
            print("Враг уже повержен")
            return
        self.current_weapon.owner = self
        self.current_weapon.hit(target)

    def heal(self, amount):
        self.hp += amount
        if self.hp > 200:
            self.hp = 200
        print(f"Пополнены, теперь здоровье {self.hp}")

weapon1 = Weapon("Короткий меч", 5, 1)
weapon2 = Weapon("Длинный меч", 7, 2)
weapon3 = Weapon("Лук", 3, 10)
weapon4 = Weapon("Лазерная орбитальная пушка", 1000, 1000)
princess = BaseCharacter(100, 100, 100)
archer = BaseEnemy(50, 50, weapon3, 100)
armored_swordsman = BaseEnemy(10, 10, weapon2, 500)
archer.hit(armored_swordsman)
armored_swordsman.move(10, 10)
print(armored_swordsman.get_coords())
main_hero = MainHero(0, 0, "Король Артур", 200)
main_hero.hit(armored_swordsman)
main_hero.next_weapon()
main_hero.add_weapon(weapon1)
main_hero.hit(armored_swordsman)
main_hero.add_weapon(weapon4)
main_hero.next_weapon()
main_hero.hit(armored_swordsman)
main_hero.next_weapon()
main_hero.hit(princess)
main_hero.hit(armored_swordsman)
```

```python
class MailServer:
    servers = {}

    def __init__(self, name):
        self.name = name
        self.mailboxes = {}
        MailServer.servers[name] = self

    @classmethod
    def get_server(cls, name):
        return cls.servers.get(name)

    @classmethod
    def available_servers(cls):
        return list(cls.servers.keys())

    def add_user(self, user):
        if user not in self.mailboxes:
            self.mailboxes[user] = []

    def receive_mail(self, user, message):
        self.add_user(user)
        self.mailboxes[user].append(message)

    def fetch_mail(self, user):
        self.add_user(user)
        mails = self.mailboxes[user][:]
        self.mailboxes[user] = []
        return mails if mails else ["Почтовый ящик пуст"]

class MailClient:
    def __init__(self, server, user):
        if isinstance(server, str):
            server = MailServer.get_server(server)
        if not server:
            print("Сервер не найден\n")
            return
        self.server = server
        self.user = user
        self.server.add_user(user)

    def receive_mail(self):
        return self.server.fetch_mail(self.user)

    def send_mail(self, server_name, user, message):
        server = MailServer.get_server(server_name)
        if not server:
            print(f"Сервер '{server_name}' не найден. Невозможно отправить письмо.\n")
            return
        server.receive_mail(user, f"From {self.user}@{self.server.name}: {message}")

    @staticmethod
    def list_servers():
        return MailServer.available_servers()

MailServer("mail1")
MailServer("mail2")

print(MailClient.list_servers())

Alexandr = MailClient("mail1", "alex")
Anatoliy = MailClient("mail2", "anatoliy")
Scarlett = MailClient("mail3", "scarlett")
Scarlett = MailClient("mail1", "scarlett")

Alexandr.send_mail("mail2", "anatoliy", "Привет, Анатолий!")
Anatoliy.send_mail("mail1", "alex", "Поставьте 25 баллов за лабораторную, пожалуйста!")
Anatoliy.send_mail("mail1", "alex", "Ну, пожалуйста...")

Scarlett.send_mail("mail3", "charlie", "Тут кто-нибудь есть?")
print(Alexandr.receive_mail())
print(Alexandr.receive_mail())
print(Anatoliy.receive_mail())
```

```python
class File:
    def __init__(self, name):
        self.name = name
        self.content = ""

    def write(self, data):
        self.content = data

    def read(self):
        return self.content


class Directory:
    def __init__(self, name):
        self.name = name
        self.subdirs = {}
        self.files = {}

    def get_dir(self, path_parts, create=False):
        if not path_parts or path_parts == ['']:
            return self
        current = self
        for part in path_parts:
            if part not in current.subdirs:
                if create:
                    current.subdirs[part] = Directory(part)
                else:
                    raise FileNotFoundError(f"Директория '{part}' не найдена")
            current = current.subdirs[part]
        return current

    def list(self):
        return sorted(list(self.subdirs.keys()) + list(self.files.keys()))

    def add_file(self, name, content):
        self.files[name] = File(name)
        self.files[name].write(content)

    def get_file(self, name):
        if name not in self.files:
            raise FileNotFoundError(f"Файл '{name}' не найден")
        return self.files[name]


class FileSystem:
    def __init__(self):
        self.root = Directory("")

    def _split_path(self, path):
        parts = path.strip("/").split("/")
        if parts == ['']:
            return []
        return parts

    def mkdir(self, path):
        parts = self._split_path(path)
        self.root.get_dir(parts, create=True)

    def ls(self, path):
        parts = self._split_path(path)
        dir = self.root.get_dir(parts)
        return dir.list()

    def write(self, path, content):
        parts = self._split_path(path)
        *dir_parts, file_name = parts
        dir = self.root.get_dir(dir_parts, create=True)
        dir.add_file(file_name, content)

    def read(self, path):
        parts = self._split_path(path)
        *dir_parts, file_name = parts
        dir = self.root.get_dir(dir_parts)
        return dir.get_file(file_name).read()



fs = FileSystem()
fs.mkdir("aaa/bbb/ccc")
fs.mkdir("aaa/ccc")
fs.write("aaa/bbb/1.txt", "hello")
fs.write("aaa/ccc/2.txt", "world")
print(fs.ls(""))
print(fs.ls("aaa"))
print(fs.ls("aaa/ccc"))
print(fs.ls("aaa/bbb"))
print(fs.read("aaa/bbb/1.txt"))
fs.write("aaa/bbb/1.txt", "Умоляю, поставьте 25 баллов")
print(fs.read("aaa/bbb/1.txt"))
# print(fs.ls("aaa/bbc"))
```
