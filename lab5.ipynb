№1 (2 балла) 
При помощи функций filter и map выполните задания: 
1. Выведите все элементы списка, которые меньше 5. 
2. Выведите все элементы списка, разделенные на два. 
3. Выведите результат деления на два всех элементов списка, которые больше 17. 
4. Напишите программу, которая подсчитает и выведет сумму квадратов всех двузначных 
чисел, делящихся на 9. При решении задачи используйте комбинацию функций filter, map, 
sum. Обратите внимание: на 9 должно делиться исходное двузначное число, а не его 
квадрат. 

```python
lst = [2, 4, 6, 9, 18, 20, 27, 36]

less_than_5 = list(filter(lambda x: x < 5, lst))
print("Элементы меньше 5:", less_than_5)

divided_by_2 = list(map(lambda x: x / 2, lst))
print("Элементы, разделённые на два:", divided_by_2)

greater_than_17_divided_by_2 = list(map(lambda x: x / 2, filter(lambda x: x > 17, lst)))
print("Элементы больше 17, разделённые на два:", greater_than_17_divided_by_2)

sum_of_squares = sum(map(lambda x: x**2, filter(lambda x: 10 <= x <= 99 and x % 9 == 0, range(10, 100))))
print("Сумма квадратов двузначных чисел, делящихся на 9:", sum_of_squares)
```

№2 (1 балл)  
Напишите функцию генератор factorials(n), генерирующую последовательность факториалов 
натуральных чисел.

```python
def factorials(n):
    fact = 1
    for i in range(1, n + 1):
        fact *= i
        yield fact
print("Факториалы чисел от 1 до 7:")
for factorial in factorials(7):
    print(factorial, end=' ')
```

№3 (1 балл)  
Напишите функцию генератор square_fibonacci(n), генерирующую последовательность квадратов 
чисел Фибоначчи.

```python
def square_fibonacci(n):
    a, b = 1, 1
    for _ in range(n):
        yield a ** 2
        a, b = b, a + b

print("Квадраты чисел Фибоначчи (первые 7):")
for number in square_fibonacci(7):
    print(number, end=' ')
```

Дополнительные модули использовать нельзя.

```python
def russian_alphabet():
    for code in range(ord('а'), ord('я') + 1):
        yield chr(code)

for letter in russian_alphabet():
    print(letter, end=' ')
```

```python
def arithmetic_operation(operation):
    if operation == '+':
        return lambda x, y: x + y
    elif operation == '-':
        return lambda x, y: x - y
    elif operation == '*':
        return lambda x, y: x * y
    elif operation == '/':
        return lambda x, y: x / y
    else:
        raise ValueError("Недопустимая операция")

operation = arithmetic_operation('+')
print(operation(1, 4))
```

```python
def same_by(characteristic, objects):
    if not objects:
        return True
    first_value = characteristic(objects[0])
    for object in objects[1:]:
        if characteristic(object) != first_value:
            return False
    return True


values = [0, 2, 10, 6]
characteristic = lambda x: x % 2

if same_by(characteristic, values):
    print('same')
else:
    print('different')

values = [1, 2, 3, 4]
if same_by(characteristic, values):
    print('same')
else:
    print('different')
```

```python
def print_operation_table(operation, num_rows=9, num_columns=9):
    for i in range(1, num_rows + 1):
        row = [operation(i, j) for j in range(1, num_columns + 1)]
        print(*row)

print_operation_table(lambda x, y: x * y)
print("-=-=-=-=-=-=-=-=-=-=-")
print_operation_table(lambda x, y: x * y, 5)
```

```python
def ask_password(login, password, success, failure):
    vowels = set('aeiouy')
    login = login.lower()
    password = password.lower()

    login_consonants = [char for char in login if char not in vowels]
    password_consonants = [char for char in password if char not in vowels]

    password_vowel_count = sum(1 for char in password if char in vowels)

    wrong_vowels = password_vowel_count != 3
    wrong_consonants = password_consonants != login_consonants

    if not wrong_vowels and not wrong_consonants:
        success(login)
    elif wrong_vowels and wrong_consonants:
        failure(login, "Everything is wrong")
    elif wrong_vowels:
        failure(login, "Wrong number of vowels")
    else:
        failure(login, "Wrong consonants")

def main(login, password):
    def success(lgn):
        print(f"Привет, {lgn}!")
    def failure(lgn, msg):
        print(f"Кто-то пытался притвориться пользователем {lgn}, но в пароле допустил ошибку: {msg.upper()}.")

    ask_password(login, password, success, failure)
    
main("anastasia", "nsyatos")
main("eugene", "aanig")
ask_password("anastasia", "nsyatos", lambda login: print("super"), lambda login, err: print("bad"))
```

```python
def check_password(password):
    def decorator(func):
        def wrapper(*args, **kwargs):
            user_input = input("Введите пароль: ")
            if user_input == password:
                return func(*args, **kwargs)
            else:
                print("В доступе отказано")
                return None
        return wrapper
    return decorator

@check_password('password')
def make_burger(typeofMeat, withOnion=False, withTomato=True):
    print(f"Бургер с {typeofMeat}, лук: {withOnion}, помидор: {withTomato}")

make_burger('говядина', withOnion=True)
```

```python
def cached(func):
    cache = {}
    def wrapper(n):
        if n in cache:
            return cache[n]
        result = func(n)
        cache[n] = result
        return result
    return wrapper

# @cached
def fib(n):
    if n == 1 or n == 2:
        return 1
    else:
        return fib(n - 1) + fib(n - 2)

for i in range(1, 40):
    if i % 10 == 0:
        print('\n')
    print(fib(i), end=' ')
```
