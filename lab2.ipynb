Пользователь (или несколько пользователей за одним компьютером) вводит слова. Начиная  
со второго введённого слова, программа проверяет, совпадает ли первая буква  
свежевведённого слова с последней буквой предыдущего. Если да, то программа работает  
дальше (считывает очередное слово). Если нет — выводит последнее на этот момент  
введённое слово и завершает работу. 

```python
words = []
words.append(input("Введите первое слово: "))
words.append(input("Введите следующее слово: "))
i = 0

while True:
    if words[i][-1] == words[i + 1][0]:
        words.append(input("Введите следующее слово: "))
    else:
        print(words[i + 1])
        break
    i += 1
```

Изобразите извилистый спуск улитки по стене по заданной траектории.  
Вводится одна строка. Начальный символ этой строки — символ рисования пути улитки.  
Далее следует последовательность символов “<”, “>” и “V”, которые означают движение  
улитки, соответственно, влево, вправо и вниз.  
Гарантируется, что по любой горизонтали движение происходит только в одну сторону  
(иными словами, между идущими в любом порядке “<” и “>” всегда есть хотя бы один “V”)  
и что путь не пройдёт левее начального положения.  
Выводятся несколько строк, состоящих из символов рисования пути и пробелов. Верхний   
левый угол соответствует начальному положению улитки. Каждый последующий символ  
входной строки означает, что нужно дорисовать символ рисования пути левее, правее или  
ниже предыдущего. Пробелы используются для отступов. 

@VVV>>>>>>>>V<<VVVV<<<VV>>>  
.>>>VV<<V

```python
way = input("Введите путь улитки (“<”, “>” и “V”): ")
horizontal = 0
spaces = 0
symbol = way[0]
step = 1
back = 0
left = 1
right = 0

for step in range(len(way) - 1):
    
    if way[step] == '>':
        back = 0
        horizontal += 1
        if (way[step] != way[step + 1]) or (step + 1 == len(way) - 1):
            print(' ' * (right), symbol * (horizontal-right), sep="")
            right = 0
            
    elif way[step] == 'V':
        back = 0
        spaces = horizontal - 1
        if (way[step - 1] == 'V') or ((way[step - 1] != '<') and (way[step - 1] != '>')):
            print(' ' * spaces, symbol, sep="")
        elif step + 1 == len(way) - 1:
            print(' ' * spaces, symbol, sep="")
            
    elif way[step] == '<':
        back += 1
        horizontal -= 1
        if horizontal - 1 < 0:
            horizontal = 1
            print("Некуда идти!")
        elif (way[step] != way[step + 1]) or (step + 1 == len(way) - 1):
            spaces = horizontal - back + left
            left += 1
            right += 1
            back += 1
            print(' ' * spaces, symbol * back, sep="")
        
```

середину и будем тянуть. Если в слове нечётное количество букв, на верхушке окажется  
одна буква. А вот если чётное, то тянуть будем за две средние буквы, они и окажутся  
наверху. Напишите программу, которая покажет, как будет выглядеть растянутое слово.  
При решении задачи можно использовать только циклы и операции над строками и  
числами. 

```python
import math

word = input("Введите слово: ")
half = math.ceil(len(word) / 2) - 1
mid_spaces = 0

if len(word) % 2 == 0:
    for i in range(half + 1):
        print(" " * (half - i) + word[half - i] + " " * mid_spaces +
              ("" if i == 0 and len(word) % 2 != 0 else word[half + i + 1]))
        mid_spaces += 2


else:
    for i in range(half + 1):
        print(" " * (half - i) + word[half - i] + " " * mid_spaces +
              ("" if i == 0 and len(word) % 2 != 0 else word[half + i]))
        mid_spaces += 2
        
#Чтобы не дублировать цикл можно засунть его в функцию
```

```python
def get_sign(number):
    if number == 0:
        return 1.0
    return number / abs(number)


def find_a_pair(numbers_list: list):
    for i, element in enumerate(numbers_list):
        if i + 1 < len(numbers_list):
            if get_sign(element) == get_sign(numbers_list[i + 1]):
                print(element, numbers_list[i + 1])
                break


find_a_pair([0, 0, 1, 3, -4, 4, -6, 3, 7, -8, 0, 1])
find_a_pair([-1, 0, -4, -2, 0])
```

(1 балл)

```python
def new_sentence(nums, sentence: str):
    nums = list(map(lambda x: int(x) - 1, nums.split(" ")))
    sentence_list = sentence.lower().split(" ")

    answer = ''
    for num in nums:
        answer += sentence_list[num] + " "
    return answer.capitalize()[:-1]


print(new_sentence("4 3 1", "Буря мглою небо кроет"))
print(new_sentence("3 2 1", "Ты и я"))
```

```python
import math

coords = []
for i in range(8):
    coords.append(input(f'Место {i + 1} ферзя: ').split(' '))

coords = list(map(lambda c: [int(c[0]), int(c[1])], coords))


def queen():
    for coord in coords:
        x, y = coord
        copy = coords.copy()
        copy.remove(coord)
        for queen_coord in copy:
            queen_x, queen_y = queen_coord
            if x == queen_x or y == queen_y:
                return True
    return False

print("YES") if queen() else print("NO")
```

№ 15  
(2 балла)

Кто последний? Я - Кузнецов.  
Кто последний? Я - Поливанов.  
Следующий!  
Я только спросить! Я - Иванова.

```python
quantity = int(input("Введите количество фраз: "))
patient_list = []

for i in range(quantity):
    phrase = input()
    if "Кто последний?" in phrase or "Я только спросить!" in phrase:
        surname = (phrase.split("-")[1]
                   .replace(".", "")
                   .replace(" ", ""))
        if "Кто последний?" in phrase:
            patient_list.append(surname)
        else:
            patient_list.insert(0, surname)
    elif "Следующий!" in phrase:
        if patient_list:
            print(f"Заходит {patient_list.pop(0)}!")
        else:
            print("В очереди никого нет")
```
