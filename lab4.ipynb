При помощи синтаксиса списочных выражений выполните задания: 
1. Пусть дан список my_list. Создайте новый список из элементов my_list, которые меньше 5. 
2. Пусть дан список my_list. Создайте новый список из элементов my_list, деленных на 2. 
3. Пусть дан список my_list. Создайте новый список из элементов my_list, которые больше 17, 
умножив их на 2. 
4. Создайте список, включающий числа от 0 до введённого пользователем, возведённые в 
квадрат. 
5. Используя списочное выражение и метод split, составьте список из введённых чисел, 
записанных на одной строке без указания заранее их количества; затем выведите их 
квадраты также на одной строке. 
6. Используя списочное выражение и метод split, составьте список из введённых чисел, 
записанных на одной строке без указания заранее их количества; затем выведите на одной 
строке только те квадраты нечетных чисел, которые не заканчиваются на цифру 9. 
Постарайтесь решить данную задачу в одну строку.

1 2 3 4 5 6 7 8 9 12 45 23 25 52 43 18 35 29 49 19

```python
my_list = input().split()
first_list = [number for number in my_list if int(number) < 5]
second_list = [int(number) / 2 for number in my_list]
third_list = [int(number) * 2 for number in my_list if int(number) > 17]
fifth_list = [int(number) ** 2 for number in my_list]
sixth_list = [int(number) ** 2 for number in my_list if (int(number) % 2 == 1) and (int(number[-1]) != 9)]
#sixth_list = [int(number) ** 2 for number in my_list if (int(number) % 2 == 1) and (int(str(int(number) ** 2)[-1]) != 9)]

up_to_number = int(input("Введите до какого числа будет расчёт"))
fourth_list = [int(number) ** 2 for number in range(up_to_number + 1)]

print(first_list, second_list, third_list, fourth_list, sep='\n')
print(*fifth_list, sep=' ')
print(*sixth_list, sep=' ')
```

```python
words_list = {
    1: 'один',           2: 'два',           3: 'три',
    4: 'четыре',         5: 'пять',          6: 'шесть',
    7: 'семь',           8: 'восемь',        9: 'девять',
    10: 'десять',        11: 'одиннадцать',  12: 'двенадцать',
    13: 'тринадцать',    14: 'четырнадцать', 15: 'пятнадцать',
    16: 'шестнадцать',   17: 'семнадцать',   18: 'восемнадцать',
    19: 'девятнадцать',  20: 'двадцать',     30: 'тридцать',
    40: 'сорок',         50: 'пятьдесят',    60: 'шестьдесят',
    70: 'семьдесят',     80: 'восемьдесят',  90: 'девяносто'}


def number_to_words(n):
    if (n <= 20) and (n > 0):
        return words_list[n]
    elif (n > 20) and (n <= 99):
        if n % 10 == 0:
            return words_list[n // 10]
        else:
            return f'{words_list[(n // 10) * 10]} {words_list[n % 10]}'
    return 'Ошибка. Это должно быть число от 1 до 99!'


print(number_to_words(int(input())))
```

```python
def bracket_check(test_string: str):
    check_stack = []
    for symbol in test_string:
        if symbol == '(':
            check_stack.append(symbol)
        else:
            if check_stack[-1] != '(' or not check_stack:
                return 'NO'
            check_stack.pop()
    if not check_stack:
        return 'YES'
    else:
        return 'NO'


print(bracket_check(input()))
```

```python
person_and_his_friends = dict()


def add_friends(name_of_person:str, list_of_friends:list):
    #person_and_his_friends[name_of_person] = list_of_friends
    
    if name_of_person not in person_and_his_friends:
        person_and_his_friends[name_of_person] = list()
    
    for friend in list_of_friends:
        person_and_his_friends[name_of_person].append(friend)


def are_friends(name_of_person1:str, name_of_person2:str):
    
    if name_of_person2 in person_and_his_friends[name_of_person1]:
        return True
    return False


def print_friends(name_of_person:str):
    
    print(*sorted(person_and_his_friends[name_of_person]))


add_friends("Алла", ["Марина", "Иван"])
print(are_friends("Алла", "Мария"))
add_friends("Алла", ["Мария"])
print(are_friends("Алла", "Мария"))
print_friends("Алла")
```

```python
def continue_fibonacci_sequence(sequence, n):
    for i in range(n):
        if len(sequence) < 2:
            return 'Длина списка должна быть минимум 2!'
        next_element = sequence[-1] + sequence[-2]
        sequence.append(next_element)
    print(*sequence)

continue_fibonacci_sequence([1, 1, 2, 3, 5], 3)
```

```python
def solve(*coefficients):
    roots = []
    if len(coefficients) == 1:
        c = coefficients
        return [] if c != 0 else ["Любые"]
    
    elif len(coefficients) == 2:
        b, c = coefficients
        
        if b == 0 and c == 0: return ["Любые"]
        
        elif b == 0 and c != 0: return []
        
        else:
            roots.append(-c / b)
            return roots
        
    elif len(coefficients) == 3:
        a, b, c = coefficients
        roots = []
    
        if a == 0 and b == 0 and c == 0: return ["Любые"]
        
        elif a == 0 and b == 0: return ["Любые"]
        
        elif c == 0:
            roots.append(0)
            roots.append((-1 * b) // a)
            return roots
        
        elif a == 0: roots.append(-c / b)
        
        else:
            d = b ** 2 - 4 * a * c
            
            if d > 0:
                roots.append((-b + d // d) // 2 * a)
                roots.append((-b - d // d) // 2 * a)
                
            if d == 0: roots.append(-b // 2 * a)
            
        return roots
    else: return None


print(sorted(solve(1, 2, 1)))
print(sorted(solve(1, -3, 2)))
```
