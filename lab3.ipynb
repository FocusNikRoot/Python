```python
quantity = int(input())
surname_dict = dict()
count = 0

for i in range(quantity):
    surname = input()
    if surname not in surname_dict:
        surname_dict[surname] = 1
    else:
        surname_dict[surname] += 1


for value in surname_dict.values():
    if value > 1:
        count += value


print(count)
```

№ 8  
(2 балла)

```python
quantity = int(input())
synonyms_dict = dict()


def get_key(dict, word):
    for key, value in dict.items():
        if value == word:
            return key


for i in range(quantity):
    synonyms = input("Введите 2 синонима через пробел: ").split(" ")
    synonyms_dict[synonyms[0]] = synonyms[1]


word = input("Слово для поиска синонима: ")
if word in synonyms_dict:
    print(synonyms_dict[word])
elif word in synonyms_dict.values():
    print(get_key(synonyms_dict, word))
else:
    print(f'В словаре нет слова {word}!')
```

№ 9  
(2 балла)

```python
quantity = int(input())
candidate_votes_count = dict()


def print_candidate_votes(dict):
    for key, value in sorted(dict.items()):
        print(key, value, end='\n')


for i in range(quantity):
    candidate, votes = input().split()
    if candidate in candidate_votes_count:
        candidate_votes_count[candidate] += int(votes)
    else:
        candidate_votes_count[candidate] = int(votes)


print_candidate_votes(candidate_votes_count)
```

```python
operations_dict = {"write": "W", "read": "R", "execute": "X"}
permissions_dict = dict()

files_quantity = int(input())
for i in range(files_quantity):
    create_file = input().split()
    permissions_dict[create_file[0]] = create_file[1:]

call_quantity = int(input())

for i in range(call_quantity):
    operation, file = input().split()
    if operations_dict[operation] in permissions_dict[file]:
        print("OK")
    else:
        print("Access denied")
```
